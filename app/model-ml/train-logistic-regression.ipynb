# Bài 1 · Accuracy 90% có đủ tốt không?

Giả sử trong **100 ảnh** có **10 ảnh số 7**. Một mô hình luôn trả lời “chữ số khác”
đoán đúng 90 ảnh, nên **accuracy đạt 90%**.

Tuy nhiên, mô hình tìm được **0 trên 10 ảnh số 7**, tức **recall của lớp số 7 bằng 0%**.

> **Điểm cần nhớ:** Accuracy cao chưa đủ để kết luận mô hình tìm được ảnh số 7.

Trong bài này, bạn sẽ kiểm tra điều đó trên dữ liệu thật từ bộ **Digits** của scikit-learn.


## Bước 1 · Chia dữ liệu 60/20/20

**Mục tiêu:** Nhận diện ảnh chữ số **7** so với các chữ số còn lại.

- **Đặc trưng `X`:** Giá trị pixel được chia cho `16.0` để đưa về khoảng `[0, 1]`.
- **Nhãn `y`:** `1` nếu ảnh là số 7, `0` nếu là chữ số khác.

| Tập dữ liệu | Tỷ lệ | Mục đích |
| :--- | ---: | :--- |
| Train | 60% | Huấn luyện mô hình |
| Validation | 20% | Đánh giá và lựa chọn mô hình |
| Test | 20% | Đánh giá cuối cùng |

Chia dữ liệu theo hai lượt: giữ **60%** cho train, rồi chia đôi **40%** còn lại
thành validation và test.

> `stratify` giúp giữ tỷ lệ hai lớp gần tương đương giữa các tập.
> `random_state=42` giúp tái lập kết quả khi chạy lại.


```python
import numpy as np
from sklearn.datasets import load_digits
from sklearn.metrics import accuracy_score, confusion_matrix, recall_score
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression

digits = load_digits()
X = digits.data / 16.0
y = (digits.target == 7).astype(int)

# Lấy 60% làm train, giữ lại 40% để chia tiếp
X_train, X_rest, y_train, y_rest = train_test_split(
    X, y, test_size=0.4, random_state=42, stratify=y
)

# Chia đôi phần 40% thành validation và test
X_val, X_test, y_val, y_test = train_test_split(
    X_rest, y_rest, test_size=0.5, random_state=42, stratify=y_rest
)

print("Train:", X_train.shape)
print("Validation:", X_val.shape)
print("Test:", X_test.shape)
print("Tỷ lệ ảnh số 7 trong validation:", y_val.mean())

model = LogisticRegression(max_iter=1000)
model.fit(X_train, y_train)
p_val = model.predict_proba(X_val)[:, 1]
print("Xác suất số 7:", p_val[:5].round(3))
print("Nhãn thật: ", y_val[:5])
```
