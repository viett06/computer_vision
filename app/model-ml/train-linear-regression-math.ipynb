### Tạo dữ liệu và viết hàm MSE

```python
import numpy as np

X = np.array([[0.0], [1.0], [2.0]])  # 3 mẫu, mỗi mẫu 1 đặc trưng
y = 3 * X[:, 0] + 2                 # [2.0, 5.0, 8.0]


def mean_squared_error(y_true, y_pred):
    error = y_pred - y_true
    return np.mean(error ** 2)


print("X:", X.shape)  # (3, 1)
print("y:", y.shape)  # (3,)
```

### Tính đúng một bước cập nhật

```python
def linear_gradients(X, y_true, y_pred):
    error = y_pred - y_true
    dw = 2 * (X.T @ error) / len(y_true)
    db = 2 * error.mean()
    return dw, db


w = np.zeros(X.shape[1])
b = 0.0
learning_rate = 0.1

y_pred = X @ w + b
dw, db = linear_gradients(X, y, y_pred)

w -= learning_rate * dw
b -= learning_rate * db

print("Gradient:", dw, db)  # [-14.] -10.0
print("w, b mới:", w, b)    # [1.4] 1.0
print("MSE mới:", mean_squared_error(y, X @ w + b))  # ≈ 7.4667
```

### Lặp bước cập nhật trong fit()

```python
class LinearRegression:
    def __init__(self, learning_rate=0.1, n_steps=1000):
        self.learning_rate = learning_rate
        self.n_steps = n_steps

    def fit(self, X, y):
        self.w = np.zeros(X.shape[1])
        self.b = 0.0
        self.loss_history = []

        for _ in range(self.n_steps):
            y_pred = self.predict(X)
            self.loss_history.append(mean_squared_error(y, y_pred))

            dw, db = linear_gradients(X, y, y_pred)
            self.w -= self.learning_rate * dw
            self.b -= self.learning_rate * db

        return self

    def predict(self, X):
        return X @ self.w + self.b
```

### Train và dự đoán trên hai mẫu mới

```python
model = LinearRegression().fit(X, y)

X_test = np.array([[0.5], [1.5]])
y_test = 3 * X_test[:, 0] + 2
y_test_pred = model.predict(X_test)

print("w:", model.w.round(4))                          # [3.]
print("b:", round(model.b, 4))                         # 2.0
print("MSE ban đầu:", model.loss_history[0])            # 31.0
print("MSE train cuối:", mean_squared_error(y, model.predict(X)))
print("Dự đoán test:", y_test_pred.round(4))             # [3.5 6.5]
print("Nhãn thật:", y_test)                             # [3.5 6.5]
```

### Stochastic Gradient Descent (SGD)

```python
rng = np.random.default_rng(42)

w_sgd = np.zeros(X.shape[1])
b_sgd = 0.0
learning_rate = 0.1
n_epochs = 10
for epoch in range(n_epochs):
     for i in rng.permutation(len(y)):
         x_i = X[i]
         y_i = y[i]
         y_pred_i = x_i @ w_sgd + b_sgd
         error = y_pred_i - y_i

         dw = 2 * error * x_i
         db = 2 * error

         w_sgd -= learning_rate* dw
         b_sgd -= learning_rate * db
     # Sau một epoch, tính MSE trên toàn tập train để theo dõi
     train_predictions = X @ w_sgd + b_sgd
     mse = mean_squared_error(y, train_predictions)
     print(f"Epoch {epoch + 1:2d} | MSE = {mse:.6f}")

print("w:", w_sgd.round(4))
print("b:", round(b_sgd, 4))
```
